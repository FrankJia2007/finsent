```python
import pandas as pd
from finsent.data import ID2LABEL, PROCESSED_DIR     # ← 把路径也 import 进来

train = pd.read_parquet(PROCESSED_DIR / "train.parquet")   # ← 不再依赖 cwd

# 1) 类别分布：全集 vs AllAgree 子集
print(train["label"].map(ID2LABEL).value_counts(normalize=True))
print(train[train.all_agree]["label"].map(ID2LABEL).value_counts(normalize=True))

# 2) 句子长度（按空格数词）
train["n_words"] = train["sentence"].str.split().str.len()
print(train["n_words"].describe())        # 看均值、最大值 → 决定 max_length

# 3) 每类随机看 5 句，亲眼理解「金融视角」的标注标准
for lab, name in ID2LABEL.items():
    print(f"\n=== {name} ===")
    for s in train[train.label == lab].sample(5, random_state=0)["sentence"]:
        print("-", s)

# 4) 难例：不在 AllAgree 里的句子（标注员有分歧）
print(train[~train.all_agree].sample(10, random_state=1)[["sentence", "label"]])
```
